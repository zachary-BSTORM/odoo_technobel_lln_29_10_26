#### Exercice 1 Le mot le plus long FACILE
- ~ 8 min
- Signature : def mot_le_plus_long(phrase: str) -> str:

#### Énoncé
- Écrire une fonction qui reçoit une phrase et renvoie son mot le plus long.
Les signes de ponctuation et symboles ne font pas partie des mots : seuls les lettres et les
chiffres comptent.

- Les mots sont séparés par des espaces. En cas d'égalité de longueur,
renvoyer le premier mot trouvé.

- La phrase contient toujours au moins un mot.

```python
def mot_le_plus_long(phrase: str) -> str:
    if len(phrase) < 1:
        return ""
    
    best = ""
    for m in phrase.split():
        result = "".join(c for c in m if c.isalnum())
        if len(result) > len(best):
            best = result
    return best

values = [
    "le chat&!! dort" ,
    "Python est génial !!!" ,
    "a bb ccc"
]

for v in values:
    print(mot_le_plus_long(v))
```

#### Exercice 2 Lettres encadrées FACILE
- ~ 10 min
- Signature : def lettres_encadrees(s: str) -> str:

#### Énoncé
- La chaîne s contient des lettres et les symboles + et =.
Renvoyer la chaîne "true" si chaque lettre est immédiatement précédée et suivie d'un +,
sinon renvoyer "false".

- La chaîne contient toujours au moins une lettre. Un même + peut servir à deux lettres voisines.

```python
def lettres_encadrees(s: str) -> str:
    for i in range(len(s)):
        if s[i].isalpha():

            if i == 0 or i == len(s) - 1:
                return "false"

            if s[i -1] != "+" and s[i + 1] != "+":
                return "false"
    return "true"

values = [
    "+d+=+s+" ,
    "f++d+" ,
    "++a+b+"
]

for v in values:
    print(lettres_encadrees(v))
```

#### Exercice 3 Intersection de deux listes FACILE / MOYEN
- ~ 12 min
- Signature : def intersection(tab: list[str]) -> str:

#### Énoncé
- Le paramètre tab contient deux chaînes. Chacune représente une liste d'entiers triés par
ordre croissant, séparés par une virgule et un espace, sans doublon à l'intérieur d'une même
liste.

- Renvoyer les nombres présents dans les deux listes, dans l'ordre croissant, sous forme d'une
chaîne séparée par des virgules sans espace. S'il n'y a aucun nombre commun, renvoyer
"false".

```python
def intersection(tab: list[str]) -> str:
    # if len(tab[0]) < 1 or len(tab[1]) < 1:
    #     return "false"
    
    first = tab[0].split(", ")
    second = set(tab[1].split(", "))

    communs = [v for v in first if v in second and v.isnumeric()]

    if not communs :
        return "false"
    return ",".join(communs)


value = ["", ""]
print(intersection(value))
```

~ 12 min
Signature

Énoncé
Renvoyer "true" si l'identifiant respecte toutes les règles suivantes, sinon "false" :
1. il contient entre 4 et 25 caractères inclus ;
2. il commence par une lettre ;
3. il ne contient que des lettres non accentuées (a-z, A-Z), des chiffres et des underscores _ ;
4. il ne se termine pas par un underscore.

```python
def identifiant_valide(nom: str) -> str:
    if not (4 <= len(nom) <= 25):
        return "false"

    if nom[-1] == "_":
        return "false"

    if not ("a" <= nom[0].lower() <= "z"):
        return "false"

    for c in nom:
        is_correct = "a" <= c.lower() <= "z"
        if not (is_correct or c.isdigit() or c == "_"):
        # if not (c.isalpha() and c.isascii() or c.isdigit() or c == "_"):
            return "false"

    return "true"


value = "u__hello_world123"
print(identifiant_valide(value))
```

#### Exercice 5 Compression d'une chaîne MOYEN
- ~ 15 min
- Signature : def compresser(s: str) -> str:
#### Énoncé
La chaîne s ne contient que des lettres minuscules et n'est jamais vide.
Renvoyer sa version compressée : chaque suite de lettres identiques et consécutives est
remplacée par le nombre de répétitions suivi de la lettre.

```python
def compresser(s: str) -> str:
    morceaux = []
    compteur = 1
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            compteur += 1
        else:
            morceaux.append(f"{compteur}{s[i - 1]}")
            compteur = 1
    morceaux.append(f"{compteur}{s[- 1]}")
    return "".join(morceaux)


value = "wwwggopp"

print(compresser(value))
```

Exercice 1 Somme des chiffres FACILE
~ 8 min
Signature
def somme_chiffres(s: str) -> int:
Énoncé
La chaîne s contient des lettres et des chiffres. Renvoyer la somme de tous les chiffres
qu'elle contient, sous forme d'entier.
Contrainte : la fonction int() est interdite. La conversion d'un caractère en valeur numérique
doit passer par son code ASCII.
Si la chaîne ne contient aucun chiffre, renvoyer 0.

```python
def somme_chiffres(s: str) -> int:
    total = 0
    for c in s:
        if "0" <= c <= "9":
            total += ord(c) - ord("0")

    return total


value = "abc"

print(somme_chiffres(value))
```

Le paramètre n est un nombre entier de minutes, positif ou nul.
Renvoyer la durée au format heures, lettre h, minutes sur deux chiffres. Le nombre
d'heures n'a pas de limite.

```python
def type_suite(nombres: list[int]) -> str:
    # ecart
    ecart = nombres[1] - nombres[0]  # récupération de l'écart
    if all(nombres[i] - nombres[i - 1] == ecart for i in range(2, len(nombres))):
        return "arithmetique"

    # rapport
    a, b = nombres[0], nombres[1]  # récupération des premières valeurs
    # [2, 6, 18, 54]
    # Si i == 2
    # (nb[i] == 18) & (nb[i - 1] == 6)
    #  ( 18 * 2 ) == (6 * 6 )
    #  (36)       == (36)
    if all(nombres[i] * a == nombres[i - 1] * b for i in range(2, len(nombres))):
        return "geometrique"

    return "aucune"

ar = [2, 4, 6, 8]
geo = [2, 6, 18, 54]
aucun = [1, 2, 4, 7]


type_suite(ar)
type_suite(geo)
type_suite(aucun)
```

Exercice 5 Parenthèses équilibrées MOYEN
~ 15 min
Signature
def parentheses_equilibrees(s: str) -> str:
Énoncé
La chaîne s peut contenir des parenthèses ( ), des crochets [ ], des accolades { } et d'autres
caractères, qui sont ignorés.
Renvoyer "true" si chaque symbole ouvrant est fermé par le symbole correspondant, dans le
bon ordre, sinon "false".

```python
def parentheses_equilibrees(s):
    paires = {")":"(","]":"[","}":"{"}
    pile = []
    for c in s:
        if c in "([{":
            pile.append(c)
        elif c in paires:
            if not pile or pile.pop() != paires[c]:
                return "false"
    return "true" if not pile else "false" 
             
    
 # ["(",]   
value = '(a[b]{c})'
print(parentheses_equilibrees(value))
```

#### Exercice 1 Premier client récurrent FACILE
- ~ 10 min
- Signature : def premier_client_recurrent(commandes: list, minimum: int) -> str:
#### Énoncé
- Chaque commande est un dictionnaire, où "client" contient le nom du client.
Écrire une fonction qui renvoie le nom du premier client qui atteint minimum commandes.
Les noms peuvent avoir été mal saisis : "Deco Addict", "deco addict" et " DECO ADDICT " sont le même
client. On renvoie le nom tel qu'il est écrit dans la première commande de ce client, sans les espaces
autour.
Si aucun client n'atteint minimum commandes, on renvoie la chaîne "-1".

```python
def premier_client_recurrent(commandes : list[dict[str:str]], minimum:int) -> str:
    compteur = {}
    name = {}
    
    for c in commandes:
        nom = c["client"].strip()
        cle = nom.lower()
        
        if cle not in name:
            name[cle] = nom
        compteur[cle] = compteur.get(cle,0) + 1
        
        if compteur[cle] == minimum:
            return name[cle]
    return "-1"
```

#### Exercice 3 Même motif FACILE
- ~ 10 min
- Signature : def meme_motif(a: str, b: str) -> str:
#### Énoncé
- Écrire une fonction qui reçoit deux chaînes et renvoie "true" si on peut remplacer chaque caractère de a
pour obtenir b, sinon "false".
Un même caractère est toujours remplacé par le même caractère, et deux caractères différents ne peuvent
pas donner le même caractère.
Le résultat est une chaîne de caractères ("true" ou "false"), pas un booléen.

```python
def meme_motif(a: str, b: str) -> str:
    if len(a) != len(b):
        return "false"
    a_to_b = {}
    b_to_a = {}
    for v_a, v_b in zip(a, b):
        if a_to_b.setdefault(v_a, v_b) != v_b or b_to_a.setdefault(v_b, v_a) != v_a:
            return "false"
    return "true"
```

#### Exercice 5 Fenêtres sans doublon FACILE +
- ~ 15 min
- Signature : def fenetres_distinctes(texte: str, k: int) -> int:
#### Énoncé
- Écrire une fonction qui compte les sous-chaînes consécutives de longueur exactement k dont tous les
caractères sont différents.
Les sous-chaînes peuvent se chevaucher. Les majuscules sont distinctes des minuscules.

```python
def fenetres_distinctes(texte: str, k: int) -> int:
    if k > len(texte):
        return 0

    compte = 0
    resultat = {}
    for i, c in enumerate(texte):
        resultat[c] = resultat.get(c, 0) + 1
        if i >= k:
            sortant = texte[i - k]
            resultat[sortant] -= 1
            if resultat[sortant] == 0:
                # del resultat[sortant]
                resultat.pop(sortant)
        if len(resultat) == k:
            compte += 1
    return compte


value = "xyzzaz"
qty = 3

print(fenetres_distinctes(value, qty))
```

#### Exercice 9 Le plus grand réservoir MOYEN
- ~ 20 min
- Signature : def plus_grand_reservoir(hauteurs: list) -> int:
#### Énoncé
- La fonction reçoit une liste d'entiers positifs. Chaque valeur est la hauteur d'une barre verticale ; les barres
sont espacées de 1. Deux barres forment avec le sol un réservoir.
La fonction renvoie la quantité d'eau maximale qu'un réservoir peut contenir : distance entre les deux
barres × hauteur de la plus petite des deux. S'il y a moins de deux barres, elle renvoie 0.

```python
def plus_grand_reservoir(hauteurs: list) -> int:
    g = 0
    d = len(hauteurs) - 1
    best = 0
    while g < d:
        best = max( best,(d - g) * min(hauteurs[g],hauteurs[d]))
        if hauteurs[g] < hauteurs[d]:
            g += 1
        else: 
            d -= 1
    return best
    
value = [1, 8, 6, 2, 5, 4, 8, 3, 7]

print(plus_grand_reservoir(value))
```

#### Exercice 6 Plus longue sous-chaîne sans répétition MOYEN
- ~ 20 min
- Signature : def plus_long_sans_repetition(texte: str) -> int:
#### Énoncé
- Écrire une fonction qui renvoie la longueur de la plus longue sous-chaîne consécutive dont aucun caractère
n'est répété.
Les espaces comptent comme des caractères. On renvoie une longueur (un entier), pas la sous-chaîne.

```python
def plus_long_sans_repetition(texte: str) -> int:
    last = {}
    start = 0
    best = 0
    for i,v in enumerate(texte):
        if v in last and last[v] >= start:
            start = last[v] + 1
        last[v] = i
        best = max(best,i - start + 1)
    return best
    
value = 'bbbbb'

print(plus_long_sans_repetition(value))
```

Exercice 12 Position d'insertion FACILE
~ 10 min
Signature
def position_insertion(nombres: list, cible: int) -> int:
Énoncé
Écrire une fonction qui reçoit une liste d'entiers distincts triée par ordre croissant et un entier cible.
Si cible est présente, renvoyer son indice. Sinon, renvoyer l'indice où il faudrait l'insérer pour que la liste
reste triée (cet indice peut valoir len(nombres)).

```python
def position_insertion(nombres: list, cible: int) -> int:
    g = 0
    d = len(nombres)
    
    while g < d:
        middle = (g + d) // 2
        if nombres[middle] < cible:
            g = middle + 1
        elif nombres[middle] > cible:
            d = middle
        else:
            return middle
    return g
print(position_insertion([1, 3, 5, 6], 7))
```
