#### vérification de chaines :

```python
print("a".isalnum()) # alphanumerique
print("a".isalpha()) # alphabetique
print("a".isnumeric()) # numerique
print("a".upper()) # majuscule
print("A".lower()) # minuscult
print("hello".title()) # première lettre (de chaque mot) en majuscule

# split() séparation 

# vérification symbole 
import string

value = "hello !"

for v in value:
    if v in string.punctuation:
        print(v)
```

- Concatenation

```python
m = "hello world"
# join ne crée la chaine que lorsqu'il à tout les élément
# contrairement au : += ou +
result = "".join(c for c in m if c.isalnum())
```

syntaxe python

```python
x = 5
value = "hello"
result_1 = 5 if x > 10 else 10 
result_2 = [e for e in value if e != ""]
result_3 = [e if e != "" else "b" for e in value]
```
