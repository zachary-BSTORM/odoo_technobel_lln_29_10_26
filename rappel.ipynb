#### Timeit permet de calculer le temps d'execution

```python
import timeit

def premier_unique_count(s: str) -> str:
    for v in s :
        if s.count(v) == 1:
            return v
    return "aucun"


def premier_unique(s):
    compte = {}
    
    for v in s:
        compte[v] = compte.get(v,0) + 1
        
    for c in s:
        if compte[c] == 1:
            return c
    return "aucun"





temps1 = timeit.timeit(
    stmt=lambda: premier_unique("abracadabraabracadabra"),
    number=5_000          # répète 5000 fois
)

temps2 = timeit.timeit(
    stmt=lambda: premier_unique_count("abracadabraabracadabra"),
    number=5_000          # répète 5000 fois
)

print(f"Temps total  : {temps1}")
print(f"Temps total  : {temps2}")
```

# Any & All

```python
result_any = any([True,False,False]) # au moins un resultat true
result_all = all([True,True,True]) # tous les résulat doivent être true

print(result_any)
print(result_all)
```

#### Dictionnaires

```python
compteur = {}
cle = "a"

# récupère ou renvoie une valeur par défaut
compteur[cle] = compteur.get(cle,0) + 1 

# renvoie la valeur et l'ajoute si absent
compteur.setdefault(cle, "defaut")

```

# Slicing

```python
liste = [1,2,3]

# le slicing peut être utilisé avant ou après le "=" 
# même fonctionnement des deux cotés [start:stop:step]
def update_liste(value):
    value[:] = [6,7,8,9]
    print(value)
    
print(liste)
update_liste(liste)
print(liste)
```
